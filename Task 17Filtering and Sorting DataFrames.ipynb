```python
import pandas as pd

data = {
    'Name': ['Ravi', 'Priya', 'Kiran', 'Sneha', 'Arjun', 'Divya'],
    'Age': [20, 22, 25, 21, 28, 24],
    'Gender': ['Male', 'Female', 'Male', 'Female', 'Male', 'Female'],
    'Marks': [75, 90, 65, 85, 95, 70]
}

df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)

# Filtering examples
print("\n1. Marks greater than 80:")
print(df[df['Marks'] > 80])

print("\n2. Age below 23:")
print(df[df['Age'] < 23])

print("\n3. Female students:")
print(df[df['Gender'] == 'Female'])

print("\n4. Age above 20 and Marks above 80:")
print(df[(df['Age'] > 20) & (df['Marks'] > 80)])

print("\n5. Age above 25 or Marks above 90:")
print(df[(df['Age'] > 25) | (df['Marks'] > 90)])

# Sorting examples
print("\n6. Marks in ascending order:")
print(df.sort_values(by='Marks'))

print("\n7. Marks in descending order:")
print(df.sort_values(by='Marks', ascending=False))

print("\n8. Age in ascending order:")
print(df.sort_values(by='Age'))

print("\n9. Names alphabetically:")
print(df.sort_values(by='Name'))

print("\n10. Multiple column sorting:")
print(df.sort_values(by=['Gender', 'Marks'],
                     ascending=[True, False]))

# Analysis
high_marks = df[df['Marks'] > 80]

print("\nAnalysis:")
print("Total students:", len(high_marks))
print("Average marks:", high_marks['Marks'].mean())
print("Highest marks:", high_marks['Marks'].max())
```
